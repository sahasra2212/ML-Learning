```python
import pandas as pd

df = pd.read_csv("melb_data.csv")

print(df["Date"].head())
```

```python
df["Date"] = pd.to_datetime(df["Date"])

print(df["Date"].head()) #output : 03/12/2016
                                   04/12/2016
```

```python
df["SoldYear"] = df["Date"].dt.year
df["SoldMonth"] = df["Date"].dt.month
df["SoldDay"] = df["Date"].dt.day

print(df[["Date", "SoldYear", "SoldMonth", "SoldDay"]].head())
```

```python
df = df.drop(columns=["Date"])

print(df.head())
```

# Date Feature Engineering

Today I learned how to extract useful information from a date column.

Instead of keeping the original date as one value, we can extract
separate features such as:

- Year
- Month
- Day

This can make date information easier for a machine learning model to use.

## What I learned

- `pd.to_datetime()` converts a column into datetime values.
- `.dt.year` extracts the year.
- `.dt.month` extracts the month.
- `.dt.day` extracts the day.
- Feature engineering means creating useful features from existing data.
- After extracting the required information, the original `Date` column can be removed.
