# Feature Scaling

Today I learned why numerical features may need to be scaled before training a machine learning model.

Different numerical features can have very different ranges.

StandardScaler transforms each feature using its mean and standard deviation.

After standardization, each feature has a mean close to 0 and a standard deviation close to 1 when calculated on the data used to fit the scaler.


```python
import pandas as pd

df = pd.read_csv("melb_data.csv")

numerical_features = df[
    ["Rooms", "Distance", "Landsize"]
]

print(numerical_features.head())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaled_data = scaler.fit_transform(numerical_features)

print(scaled_data[:5])
```

```python
scaled_df = pd.DataFrame(
    scaled_data,
    columns=numerical_features.columns
)

print("Before scaling:")
print(numerical_features.head())

print("\nAfter scaling:")
print(scaled_df.head())
```

```python
print("Means after scaling:")
print(scaled_df.mean().round(2))

print("\nStandard deviations after scaling:")
print(scaled_df.std(ddof=0).round(2))
```

## What I learned

* Numerical features can have very different ranges.
* StandardScaler standardizes features using their mean and standard deviation.
* `fit_transform()` learns the scaling parameters and transforms the data.
* Scaling is different from One-Hot Encoding: scaling adjusts numerical values, while encoding converts categories into numerical representations.
* In a real ML workflow, the scaler should be fitted on training data only to avoid data leakage.
