# One-Hot Encoding

Today I learned how categorical features can be converted into numerical values using One-Hot Encoding.

Machine learning models generally work with numerical data.

One-Hot Encoding creates a separate binary column for each category.

For example:

Type:
- h = house
- u = unit
- t = townhouse

can be converted into separate columns.

Each row gets:
- 1 if it belongs to that category
- 0 otherwise.

```python
import pandas as pd

df = pd.read_csv("melb_data.csv")

print(df[["Type", "Method"]].head())
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(handle_unknown="ignore")

encoded_data = encoder.fit_transform(df[["Type"]])

print(encoded_data.toarray())
```

```python
print("Categories:")
print(encoder.categories_)
```

## What I learned

- Categorical features contain categories rather than directly usable numerical values.
- One-Hot Encoding converts categories into binary columns.
- `OneHotEncoder` is provided by scikit-learn.
- `fit_transform()` learns the categories and transforms the data.
- `handle_unknown="ignore"` prevents errors when unseen categories appear.
- One-Hot Encoding will later be combined with other preprocessing steps.
