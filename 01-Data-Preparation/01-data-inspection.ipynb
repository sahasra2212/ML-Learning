# Dataset Inspection

Today I learned how to inspect a dataset before building a
machine learning model.

## What I learned

- `df.head()` → view the first few rows
- `df.shape` → find the number of rows and columns
- `df.info()` → check column names, data types and missing values
- `df.describe()` → get a statistical summary of numerical columns

```python
import pandas as pd

df = pd.read_csv("melb_data.csv")

print(df.shape)

df.head()

df.info()

df.describe()
```
