# Features and Target

Today I learned how to separate the target variable from the features
before training a machine learning model.

- Target (`y`) → the value we want the model to predict
- Features (`X`) → the input information used to make the prediction

```python
import pandas as pd

df = pd.read_csv("melb_data.csv")

y = df["Price"]

X = df.drop(columns=["Price"])

print("Features:")
print(X.head())

print("\nTarget:")
print(y.head())

print("\nFeatures shape:", X.shape)
print("Target shape:", y.shape)
```
