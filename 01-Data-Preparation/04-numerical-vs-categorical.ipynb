# Numerical vs Categorical Features

Today I learned how to separate numerical and categorical features.

- Numerical features contain numerical values.
- Categorical features contain categories or labels.

They need different preprocessing techniques before being given to
a machine learning model.

```python
import pandas as pd

df = pd.read_csv("melb_data.csv")

print(df.dtypes)
```

```python
numerical_features = df.select_dtypes(include=["int64", "float64"]).columns
categorical_features = df.select_dtypes(include=["str"]).columns

print("Numerical features:")
print(numerical_features)

print("\nCategorical features:")
print(categorical_features)
```

```python
print("Number of numerical features:", len(numerical_features))
print("Number of categorical features:", len(categorical_features))
```

```python
print("\nNumerical columns:")
for col in numerical_features:
    print("-", col)

print("\nCategorical columns:")
for col in categorical_features:
    print("-", col)
```

## What I learned

- Numerical features contain numerical values.
- Categorical features represent categories or labels.
- `select_dtypes()` can be used to separate them.
- Numerical and categorical features require different preprocessing.
- Numerical features will later be scaled.
- Categorical features will later be encoded.
