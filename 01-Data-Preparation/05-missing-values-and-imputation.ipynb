# Missing Values and Imputation

Today I learned how to identify missing values in a dataset
and how missing values can be handled using imputation.

Missing values can cause problems when data is given to a
machine learning model.

Common approaches include:
- Checking where values are missing
- Filling missing numerical values using the median
- Filling missing categorical values using the most frequent value

```python
import pandas as pd

df = pd.read_csv("melb_data.csv")

print("Missing values in each column:")
print(df.isnull().sum())
```

```python
missing_values = df.isnull().sum()

print("Columns containing missing values:")
print(missing_values[missing_values > 0])
```

```python
from sklearn.impute import SimpleImputer

numerical_imputer = SimpleImputer(strategy="median")
categorical_imputer = SimpleImputer(strategy="most_frequent")

print("Numerical imputation strategy: median")
print("Categorical imputation strategy: most frequent")
```

## What I learned

- `isnull().sum()` helps identify missing values.
- Missing numerical values can be filled using the median.
- Missing categorical values can be filled using the most frequent value.
- `SimpleImputer` can be used to handle missing values.
- Later, these imputers will be used inside a preprocessing pipeline.
